4. Par ou ímpar

```python
numero = int(input("Digite um número inteiro: "))
if numero % 2 == 0:
    print("O número é par.")
else:
    print("O número é ímpar.")
```

5. Situação do aluno

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
media = (nota1 + nota2) / 2

if media >= 7:
    situacao = "Aprovado"
elif media >= 5:
    situacao = "Recuperação"
else:
    situacao = "Reprovado"

print(f"Média: {media:.2f}")
print(f"Situação: {situacao}")
```

6. Maior entre três números

```python
a = int(input("Primeiro número: "))
b = int(input("Segundo número: "))
c = int(input("Terceiro número: "))

maior = a
if b > maior:
    maior = b
if c > maior:
    maior = c

print(f"Maior número: {maior}")
```

7. Contagem crescente

```python
n = int(input("Digite um número inteiro positivo N: "))
for i in range(1, n + 1):
    print(i)
```

8. Somando de 1 até N

```python
n = int(input("Digite N: "))
soma = 0
for i in range(1, n + 1):
    soma += i
print(f"Soma: {soma}")
```

9. Contando números pares

```python
pares = 0
for i in range(10):
    numero = int(input(f"Digite o {i+1}º número: "))
    if numero % 2 == 0:
        pares += 1
print(f"Foram digitados {pares} números pares.")
```

10. Soma apenas dos positivos

```python
soma_positivos = 0
for i in range(8):
    numero = int(input(f"Digite o {i+1}º número: "))
    if numero > 0:
        soma_positivos += numero
print(f"Soma dos positivos: {soma_positivos}")
```

11. Maior valor digitado

```python
maior = None
for i in range(6):
    numero = int(input(f"Digite o {i+1}º número: "))
    if i == 0 or numero > maior:
        maior = numero
print(f"Maior valor: {maior}")
```

12. Tabuada

```python
numero = int(input("Digite um número para ver a tabuada: "))
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

13. Tentativas de senha

```python
senha = int(input("Digite a senha: "))
while senha != 1234:
    print("Senha incorreta!")
    senha = int(input("Tente novamente: "))
print("Acesso permitido!")
```

14. Número secreto

```python
numero_secreto = 7
tentativas = 0
palpite = int(input("Tente adivinhar o número secreto: "))
tentativas += 1

while palpite != numero_secreto:
    if palpite < numero_secreto:
        print("Tente um número maior")
    else:
        print("Tente um número menor")
    palpite = int(input("Novo palpite: "))
    tentativas += 1

print("Parabéns! Você acertou.")
print(f"Tentativas necessárias: {tentativas}")
```

15. Analisando uma sequência de números

```python
quantidade = 0
positivos = 0
negativos = 0
soma = 0

numero = int(input("Digite um número (0 para sair): "))
while numero != 0:
    quantidade += 1
    soma += numero
    if numero > 0:
        positivos += 1
    elif numero < 0:
        negativos += 1
    numero = int(input("Digite um número (0 para sair): "))

print(f"Quantidade: {quantidade}")
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Soma: {soma}")
```
